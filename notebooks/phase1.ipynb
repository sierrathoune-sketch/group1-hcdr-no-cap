## Project Title
No Credit Approval Pending (CAP)

## Project Abstract
Lenders rely on credit history to approve loans, leaving applicants with thin or no credit files facing denials or predatory terms. Home Credit, an international consumer finance provider, serves these underbanked borrowers using alternative data to assess repayment ability. Using the Kaggle Home Credit Default Risk dataset, we aim to predict loan repayment difficulty by binary classification. The data contains 307,511 labeled applications linked to six supporting tables of credit bureau records, previous applications, and monthly payment histories in which 8% of the applicants experienced repayment difficulty. Through the project, we will aggregate each history table to one row per applicant, merge it with the application data, and train models within leakage-safe pre-processing pipelines evaluated by stratified five-fold cross-validation. Experiments will progress from logistic regression, decision tree, and random forest baselines to engineered features, tuned gradient-boosted models, ensembles, and a neural network with imbalance-aware loss functions. Performance will be measured primarily by area under the receiver operating characteristic curve (ROC-AUC), supported by precision-recall AUC, log loss, and credit industry Gini and Kolmogorov-Smirnov statistics. A baseline ROC-AUC near 0.74 and a target of about 0.78 are expected to support better-informed lending for applicants with limited credit history.

## Algorithms and Metrics

### Algorithms
We progress from interpretable baselines to high-capacity models, comparing all models on identical cross-validation folds.

| Algorithm | Implementation | Loss / criterion |
|:---|:---|:---|
| Logistic regression (baseline) | `sklearn` LogisticRegression | L2-regularized log loss |
| Decision tree, random forest | `sklearn` DecisionTreeClassifier, RandomForestClassifier | Gini impurity |
| Gradient boosting | `lightgbm` LGBMClassifier, `xgboost` XGBClassifier | Regularized binary log loss |
| Stacking ensemble | `sklearn` StackingClassifier | Log loss (LR meta-learner) |
| Multilayer perceptron | PyTorch | Weighted binary cross-entropy; focal loss |

Logistic regression provides an interpretable reference. Tree ensembles capture the feature interactions implied by weak individual correlations, and gradient boosting handles missing values natively. Focal loss down-weights easy negatives to address class imbalance.

**Logistic regression (L2-regularized log loss)**
$$J(w) = -\frac{1}{N}\sum_{i=1}^{N}\left[y_i \log \hat{p}_i + (1-y_i)\log(1-\hat{p}_i)\right] + \lambda \lVert w \rVert_2^2$$

**Decision tree / random forest (Gini impurity of a node)**
$$G = 1 - \sum_{k \in \{0,1\}} p_k^2$$

**Gradient boosting (LightGBM / XGBoost objective)**
$$\mathcal{L} = \sum_{i=1}^{N} \ell(y_i, \hat{y}_i) + \sum_{t=1}^{T} \Omega(f_t), \qquad \Omega(f) = \gamma T_{\text{leaves}} + \tfrac{1}{2}\lambda \lVert w \rVert^2$$

**Neural network (weighted binary cross-entropy)**
$$\mathcal{L}_{BCE} = -\frac{1}{N}\sum_{i=1}^{N}\left[\alpha\, y_i \log \hat{p}_i + (1-y_i)\log(1-\hat{p}_i)\right]$$

**Focal loss**
$$\mathcal{L}_{focal} = -\frac{1}{N}\sum_{i=1}^{N} \alpha_t (1-p_t)^{\gamma} \log(p_t)$$

### Metrics
With ~8% defaults, accuracy is uninformative. ROC-AUC, the Kaggle competition metric, is primary:
$$TPR=\frac{TP}{TP+FN} \qquad FPR=\frac{FP}{FP+TN} \qquad AUC=\int_0^1 TPR\,d(FPR)$$

Precision, recall, F1, and PR-AUC (average precision) assess minority-class performance; log loss (above) measures calibration:
$$P=\frac{TP}{TP+FP} \qquad R=\frac{TP}{TP+FN} \qquad F_1=\frac{2PR}{P+R} \qquad AP=\sum_n(R_n-R_{n-1})P_n$$

Domain-specific credit metrics:
$$Gini = 2\cdot AUC-1 \qquad KS=\max_t(TPR_t-FPR_t) \qquad Cost(t)=c_{FN}\,FN(t)+c_{FP}\,FP(t)$$

Gini and KS are industry standards for scorecard strength. The cost function selects the decision threshold, since a missed default costs a lender more than a rejected good applicant.

### Analysis
All metrics are reported as mean ± SD over stratified 5-fold cross-validation using leakage-safe pipelines. We track the gap between cross-validation and Kaggle leaderboard ROC-AUC to detect overfitting.

## Pipeline

All preprocessing and modeling steps are chained in a scikit-learn `Pipeline`, so every transformation is fit only on the training folds during cross-validation. This prevents data leakage and makes each experiment reproducible end to end.

### Pipeline steps

1. **Data ingestion:** load `application_train` / `application_test` (Phase 2) and the six history tables (Phase 3); downcast numeric dtypes to reduce memory.
2. **Cleaning:** replace the `DAYS_EMPLOYED` placeholder (365,243) with NaN and add an anomaly flag; drop ID columns from the feature set.
3. **Aggregation and merge (Phase 3):** aggregate each history table to one row per `SK_ID_CURR` (count, mean, max, min, sum), then left-join onto the application table, verifying the row count is unchanged.
4. **Feature engineering:** domain ratios such as credit/income, annuity/income, annuity/credit, and days employed/days birth.
5. **Preprocessing (`ColumnTransformer`):**
   - Numeric: median imputation → standard scaling
   - Categorical: most-frequent imputation → one-hot encoding (`handle_unknown="ignore"`)
6. **Model:** classifier as the final pipeline step (LR, DT, RF in Phase 2; LightGBM/XGBoost and ensembles in Phase 3; MLP in Phase 4).
7. **Validation and tuning:** stratified 5-fold CV with a fixed seed; `GridSearchCV` / `RandomizedSearchCV` over the full pipeline.
8. **Evaluation:** ROC-AUC (primary), PR-AUC, log loss, Gini, KS; results logged in an experiment table.
9. **Refit and submission:** refit the best pipeline on all training data, predict default probabilities for `application_test`, and submit to Kaggle.
